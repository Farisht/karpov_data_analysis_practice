```python
retail = retail.query('Quantity > 0').reset_index(drop=True)
retail
```

```python
customer_only_germany = retail.query('Country == "Germany"') \
    .groupby('CustomerID', as_index=False)['InvoiceNo'].nunique()

customer_only_germany
```

```python
p80 = customer_only_germany['InvoiceNo'].quantile(0.8)

p80
```

```python
customer_above_p80 = customer_only_germany[customer_only_germany['InvoiceNo'] > p80]

customer_above_p80
```

```python
germany_top = customer_above_p80.CustomerID

germany_top
```

```python
top_retail_germany =retail[
    retail['CustomerID'].isin(germany_top)
].reset_index(drop=True)

top_retail_germany
```

```python
top_retail_germany_top_description = top_retail_germany.groupby(['StockCode', 'Description'], as_index=False)['InvoiceDate'].nunique() \
    .query('StockCode != "POST"') \
    .sort_values('InvoiceDate', ascending=False)

top_retail_germany_top_description
```

```python
retail_full ['revenue'] = retail_full.Quantity * retail_full.UnitPrice

retail_full
```

```python
general_revenue = retail_full.groupby('InvoiceNo', as_index=False).agg({'revenue':'sum'}).rename(columns={'revenue':'TotalRevenue'}) \
    .sort_values('TotalRevenue', ascending=False).head(5)

general_revenue
```
