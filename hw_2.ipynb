```python
taxi_pick = taxi.groupby('borough', as_index=False) \
    .aggregate({'pickups': 'sum'}) \
    .sort_values('pickups', ascending= False)
print(taxi_pick)
```

```python
min_pickups = taxi.loc[taxi['pickups'].idxmin(), 'borough']
min_pickups

#или так min_pickups = taxi.at[taxi.pickups.idxmin(), 'borough']
```

```python
taxi_groupb = taxi.groupby(['borough', 'hday'], as_index= False) \
    .aggregate({'pickups': 'mean'})
print(taxi_groupb)
```

```python
pickups_by_mon_bor = taxi.groupby(['borough','pickup_month'], as_index=False) \
    .pickups.count() \
    .sort_values('pickups', ascending=False)

print(pickups_by_mon_bor)
```

```python
def temp_to_celcius(series):
    return (series - 32) * 5 / 9

taxi['temp_C'] = temp_to_celcius(taxi['temp'])
taxi['temp_C']
```
